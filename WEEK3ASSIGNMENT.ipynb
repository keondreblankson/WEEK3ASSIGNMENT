```python
 import pandas as pd

# Cumulative miles shown on the bike odometer at the end of each day
cumulative_miles = pd.Series([55, 120, 180, 250, 325],
                             index=["Day 1", "Day 2", "Day 3", "Day 4", "Day 5"])

print("Cumulative miles:")
print(cumulative_miles)

# Miles ridden each day
daily_miles = cumulative_miles.diff()

# The first day is the same as the first odometer reading
daily_miles.iloc[0] = cumulative_miles.iloc[0]

print("\nMiles ridden each day:")
print(daily_miles)
```
